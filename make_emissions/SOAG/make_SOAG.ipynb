So I'm also multiplying by 1.5 as that brings it into agreement with Ben's files and with CMIP6

```python
soag_recipe={'BIGALK':0.05,
             'BIGENE':0.05,
             'TOLUENE':0.15,
             'BENZENE':0.15,
             'XYLENES':0.15,
             'ISOP':0.04,
             'MTERP':0.25}
```

```python
basepath="/glade/campaign/acom/MUSICA/emis/cmip7/ne30/DRES-CMIP-BB4CMIP7-2-1_20251102/"
outpath="/glade/campaign/cgd/cas/islas/python_savs/CESM2_with_CMIP7_aer/make_emissions/SOAG/"
```

```python
soag = None
for key in soag_recipe:
    print(key)
    scalefac=soag_recipe[key]
    print(scalefac)
    dat = xr.open_mfdataset(basepath+key+'_input4MIPs*.nc')
    if soag is None:
        soag = dat.emiss*0
    soag = soag+scalefac*dat.emiss*mw[key] / mw['SOAG']
soag = soag.rename('emiss')
soag = soag*1.5
soag.to_netcdf(outpath+'SOAGx1.5_bmb_sfc_from_DRES-CMIP-BB4CMIP7-2-1_20251102.nc')
```
