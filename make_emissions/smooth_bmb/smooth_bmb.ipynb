```python
process_species = ['bc_a4','pom_a4','SO2','so4_a1','SOAG']
basepath="/glade/campaign/cgd/cas/islas/python_savs/CESM2_with_CMIP7_aer/make_emissions/bmb/sfc/CMIP7/"

for ispecies in process_species:
    dat = xr.open_dataset(basepath+ispecies+'_bmb_sfc_cmip7_for_cesm2_negspread.nc')
    emiss_sm = dat.emiss.groupby('time.month').map(lambda x: runningmean(x, 5, timeaxis='time', dropna=True) )

    # Select out the relevant time period from dat and slip in the smoothed emissions instead for output
    dat = dat.sel(time=slice(emiss_sm.time.isel(time=0), emiss_sm.time.isel(time=emiss_sm.time.size-1)))
    dat['emiss'] = emiss_sm
    dat.to_netcdf(basepath+ispecies+'_smoothed_bmb_sfc_cmip7_for_cesm2_negspread.nc')
```
